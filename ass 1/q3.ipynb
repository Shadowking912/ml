```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sn


data=pd.read_csv("CO2 Emissions.csv")
headers=data.columns
from google.colab import files
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Lasso,Ridge,SGDRegressor
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error

def linear_reggression(x,y,h):
  #splitting
  x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.2,random_state=42)
  noncategory=[3,4,7,8,9,10]

  # standardising if not already done
  st= StandardScaler()
  if h==1:
    x_train= st.fit_transform(x_train)
    x_test=st.transform(x_test)

  #linear regression
  re = LinearRegression()
  re.fit(x_train, y_train)

  #train data
  y_train_pred=re.predict(x_train)
  print("train data")
  mse=mean_squared_error(y_train, y_train_pred)
  print("mean squared error: ",mse)
  rmse=mean_squared_error(y_train, y_train_pred,squared=False)
  print("rmse:",rmse)
  r2=r2_score(y_train, y_train_pred)
  print("r2 score:",r2)
  r2_adjusted=1 - (1-r2_score(y_train, y_train_pred))*(len(y_train)-1)/(len(y_train)-x_train.shape[1])
  print("r2 adjusted",r2_adjusted)#r2_adjusted'
  mae=mean_absolute_error(y_train,y_train_pred)
  print("mae:",mae)
  print()


  #test data
  print("test data")
  y_test_pred=re.predict(x_test)
  mse2=mean_squared_error(y_test, y_test_pred)
  print("mean squared error: ",mse2)
  rmse2=mean_squared_error(y_test, y_test_pred,squared=False)
  print("rmse:",rmse2)
  r22=r2_score(y_test, y_test_pred)
  print("r2 score:",r22)
  r2_adjusted2=1 - (1-r2_score(y_test, y_test_pred))*(len(y_test)-1)/(len(y_test)-x_test.shape[1])
  print("r2 adjusted",r2_adjusted2)#r2_adjusted
  mae2=mean_absolute_error(y_test,y_test_pred)
  print("mae",mae2)
  return [mse,mse2,rmse,rmse2,r2,r22,r2_adjusted,r2_adjusted2,mae,mae2]
```

```python
def l1_reggression(x,y):
  #splitting
  x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.2)

  #standardising
  st= StandardScaler()
  x_train= st.fit_transform(x_train)
  x_test=st.transform(x_test)

  #linear regression
  re = Lasso()
  re.fit(x_train, y_train)

  #train data
  y_train_pred=re.predict(x_train)
  print("train data")
  print("mean squared error: ",mean_squared_error(y_train, y_train_pred))
  print("rmse:",mean_squared_error(y_train, y_train_pred,squared=False))
  r2=r2_score(y_train, y_train_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_train, y_train_pred))*(len(y_train)-1)/(len(y_train)-x_train.shape[1]))#r2_adjusted
  print("mae:",mean_absolute_error(y_train,y_train_pred))
  print()

  #test data
  print("test data")
  y_test_pred=re.predict(x_test)
  print("mean squared error: ",mean_squared_error(y_test, y_test_pred))
  print("rmse:",mean_squared_error(y_test, y_test_pred,squared=False))
  r2=r2_score(y_test, y_test_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_test, y_test_pred))*(len(y_test)-1)/(len(y_test)-x_test.shape[1]))#r2_adjusted
  print("mae",mean_absolute_error(y_test,y_test_pred))
```

```python
def l2_reggression(x,y):
  #splitting
  x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.2)

  #standardising
  st= StandardScaler()
  x_train= st.fit_transform(x_train)
  x_test=st.transform(x_test)

  #linear regression
  re = Ridge()
  re.fit(x_train, y_train)

  #train data
  y_train_pred=re.predict(x_train)
  print("train data")
  print("mean squared error: ",mean_squared_error(y_train, y_train_pred))
  print("rmse:",mean_squared_error(y_train, y_train_pred,squared=False))
  r2=r2_score(y_train, y_train_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_train, y_train_pred))*(len(y_train)-1)/(len(y_train)-x_train.shape[1]))#r2_adjusted
  print("mae:",mean_absolute_error(y_train,y_train_pred))
  print()


  #test data
  print("test data")
  y_test_pred=re.predict(x_test)
  print("mean squared error: ",mean_squared_error(y_test, y_test_pred))
  print("rmse:",mean_squared_error(y_test, y_test_pred,squared=False))
  r2=r2_score(y_test, y_test_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_test, y_test_pred))*(len(y_test)-1)/(len(y_test)-x_test.shape[1]))#r2_adjusted
  print("mae",mean_absolute_error(y_test,y_test_pred))
```

```python
def sgd_reggression(x,y):
  #splitting
  x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.2,random_state=12)

  #standardising
  st= StandardScaler()
  x_train= st.fit_transform(x_train)
  x_test=st.transform(x_test)

  #sgd regression
  re = SGDRegressor()
  re.fit(x_train, y_train)

  #train data
  y_train_pred=re.predict(x_train)
  print("train data")
  print("mean squared error: ",mean_squared_error(y_train, y_train_pred))
  print("rmse:",mean_squared_error(y_train, y_train_pred,squared=False))
  r2=r2_score(y_train, y_train_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_train, y_train_pred))*(len(y_train)-1)/(len(y_train)-x_train.shape[1]))#r2_adjusted
  print("mae:",mean_absolute_error(y_train,y_train_pred))
  print()

  #test data
  print("test data")
  y_test_pred=re.predict(x_test)
  print("mean squared error: ",mean_squared_error(y_test, y_test_pred))
  print("rmse:",mean_squared_error(y_test, y_test_pred,squared=False))
  r2=r2_score(y_test, y_test_pred)
  print("r2 score:",r2)
  print("r2 adjusted",1 - (1-r2_score(y_test, y_test_pred))*(len(y_test)-1)/(len(y_test)-x_test.shape[1]))#r2_adjusted
  print("mae",mean_absolute_error(y_test,y_test_pred))
```
