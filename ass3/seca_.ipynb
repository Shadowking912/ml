```python
from matplotlib import pyplot as plt
points_1=[(2,3),(4,1),(1,1)]
points_0=[(6,7),(5,6),(7,8)]
for i in points_1:
    plt.scatter(i[0],i[1],color='r')
for i in points_0:
    plt.scatter(i[0],i[1],color='b')
# plt.show()
x=[i for i in range(20)]
y=[8-x[i] for i in range(20)]
plt.ylim(0,10)
plt.xlim(0,10)
plt.plot(x,y)
plt.show()
```
